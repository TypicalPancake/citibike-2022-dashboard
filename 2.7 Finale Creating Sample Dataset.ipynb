# Import Libraries

```python
import pandas as pd
```

# Load Dataset

```python
# Load the full dataset
df = pd.read_csv("Manipulated Data/merged_citibike_weather_2022.csv", parse_dates=["date", "started_at", "ended_at"])
```

```python
print(df.head()) # Shows the first 5 rows by default
```

# Creating Random Sample of dataset

```python
df_sample = df.sample(n=80000, random_state=32)
df_sample.to_csv("sampled_citibike_weather_2022.csv", index=False)
```

```python
import os
size_mb = os.path.getsize("sampled_citibike_weather_2022.csv") / 1_000_000
print(f"Sample size: {size_mb:.2f} MB")
```
