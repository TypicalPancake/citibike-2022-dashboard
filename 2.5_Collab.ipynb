```python
import pandas as pd

# Load your merged dataset (with weather)
df = pd.read_csv(
    "merged_citibike_weather_2022.csv",
    parse_dates=["date", "started_at", "ended_at"],
    dtype={"start_station_id": "str", "end_station_id": "str"},
    low_memory=False
)
# Add a trip count column
df['trip'] = 1

# Aggregate to get trip counts between start and end stations
agg_df = df.groupby(
    ['start_station_name', 'end_station_name', 'start_lat', 'start_lng', 'end_lat', 'end_lng']
).agg({'trip': 'sum'}).reset_index()

agg_df.head()
```

```python
from keplergl import KeplerGl
# Create a map and add the aggregated trip data
m = KeplerGl(height=600, data={"Trip Flows": agg_df})
m.add_data(data=agg_df, name="Trip Flows")
m.save_to_html(file_name='kepler_map.html')
```
